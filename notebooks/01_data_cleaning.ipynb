# 01. Data Cleaning & Validation Pipeline

**Football IQ - Player Performance Prediction & Scouting System**

This notebook demonstrates the end-to-end data cleaning, handling of missing values, deduplication, sanity checking, and IQR outlier analysis.


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load raw player performance data
raw_path = os.path.join("..", "data", "raw", "players.csv")
df_raw = pd.read_csv(raw_path)
print(f"Raw dataset shape: {df_raw.shape}")
df_raw.head()
```

### 1. Data Inspection & Missing Value Audit


```python
# Checking data types and missing value counts
print("Missing values per column:")
print(df_raw.isnull().sum()[df_raw.isnull().sum() > 0])
print("\nDataset Info:")
df_raw.info()
```

### 2. Run Cleaning Pipeline from `src.data_processing`


```python
import sys
sys.path.append("..")
from src.data_processing import clean_player_data

df_clean, report = clean_player_data(df_raw)
print("Data Quality Report:")
for k, v in report.items():
    print(f" - {k}: {v}")
```

### 3. Outlier Analysis & Distribution Checks


```python
# Visualizing distributions of key metrics
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
sns.boxplot(y=df_clean["goals"], ax=axes[0], color="skyblue")
axes[0].set_title("Goals Distribution (IQR Analysis)")

sns.boxplot(y=df_clean["pass_accuracy"], ax=axes[1], color="lightgreen")
axes[1].set_title("Pass Accuracy (%) Distribution")

sns.boxplot(y=df_clean["market_value"], ax=axes[2], color="salmon")
axes[2].set_title("Market Value (€M) Distribution")
plt.tight_layout()
plt.show()
```
